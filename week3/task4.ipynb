```python
import pandas as pd
```

```python
df = pd.read_csv("major-cities-in-pakistan-by-population.csv")
```

```python
df
```

```python
print(df.columns)
```

```python
#TASK 1 the 5 smallest major cities by population

smallest_cities = df.nsmallest(5, "Population 2017 Census")

print("Five Smallest Cities:")
print(smallest_cities[["City", "Population 2017 Census"]])
```

```python
#TASK 2 which province has the highest average growth rate

average_growth = df.groupby("Province")["Change"].mean()

highest_growth = average_growth.idxmax()

print("Province with Highest Average Growth Rate:")
print(highest_growth)

print("\nAverage Growth Rate by Province:")
print(average_growth)
```

```python
#TASK 3 how many cities have a population above 1 million
cities_above_1m = df[df["Population 2017 Census"] > 1000000]

print("Number of Cities with Population Above 1 Million:")
print(len(cities_above_1m))

print("\nCities:")
print(cities_above_1m[["City", "Population 2017 Census"]])
```

```python
#TASK 4  the difference between the mean and median population, with one sentence explaining why they differ so much
mean_population = df["Population 2017 Census"].mean()

median_population = df["Population 2017 Census"].median()

difference = mean_population - median_population

print("Mean Population:", mean_population)
print("Median Population:", median_population)
print("Difference:", difference)

print("\nExplanation:")
print("The mean is much higher than the median because a few very large cities like Karachi and Lahore increase the average, while most cities have much smaller populations.")
```
